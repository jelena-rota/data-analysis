## D. Probabilities

**1. Compute the probability that a randomly selected commune with elevation over 2000 m is from the canton of Valais.**

This is conditional probability, that can be calculated as follows: P(B|A) = P(AB) / P(A)

Event A: selected commune is over 2000 m <br/>
Event B: selected commune is from the canton of Valais

P(B|A): probability that selected commune is from the canton of Valais, given that it's elevation is over 2000 m<br/>
P(AB): probability that selected commune has elevation above 2000 m and is from the canton of Valais<br/>
P(A): probabilty that selected commune has elevation above 2000 m

```python
# total number of communes
x = len(df)

# number of communes with elevation over 2000
y = len(df[df['Elevation']>2000])

# number of communes with elevation over 2000 that are from the canton of Valais
z = len(df[(df['Canton']=='VS') & (df['Elevation']>2000)])

# P(AB)
# number of communes with elevation over 2000 and from the canton of Valais divided with total number of communes
pab = z/x

# P(A)
# number of communes with elevation over 2000 divided with total number of communes
pa = y/x

# P(B|A) = P(AB)/P(A)
pba = pab / pa
pba
```

The probability that a randomly selected commune with elevation over 2000 m is from the canton of Valais is 0.51.

## E. Matrices

**1. Define a data frame matrix whose rows correspond to communes and the columns to the cantons. Fill in the matrix with 0/1 values where entry (i, j) is a 1 if the commune in row i is in the canton in column j and a 0 otherwise.**

```python
# new datafame with commune and canton columns
df1 = df[['Commune', 'Canton']].reset_index(drop=True)

# show dataframe
df1
```

```python
# new dataframe with 26 columns (canton names) and the same number of rows as df, all values are zero
df2 = pd.DataFrame(0, index=np.arange(df.shape[0]), columns=cantons)

# show dataframe
df2
```

```python
# concatenate two dataframes horizontally
df3 = pd.concat([df1, df2], axis=1)

# show dataframe
df3
```

```python
# two for loops: i going through all rows, and j going through all columns of the merged dataframe df3
# if the canton entry (i, 1) is the same as the name of the j-column, then entry (i, j) should be 1

for i in range(df3.shape[0]):
    for j in range(df3.shape[1]):
        if (df3.iloc[i, 1] == df3.columns[j]):
            df3.iloc[i,j] = 1

# show dataframe
df3
```

```python
# change index to the column 'Commune'
df3.set_index('Commune', inplace=True)

# drop the column 'Canton'
df3.drop(columns=['Canton'], inplace=True)

# show the matrix
df3
```
